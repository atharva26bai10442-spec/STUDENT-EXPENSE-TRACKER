```python
# ============================================================
# < SMARTSPEND - STUDENT EXPENSE & BUDGET TRACKER >
# ============================================================
# This program helps students:
# 1. Add expenses
# 2. View expenses
# 3. Search expenses
# 4. Edit expenses
# 5. Delete expenses
# 6. Set a monthly budget
# 7. Check budget status
# 8. Analyze expenses
# 9. Generate reports
from datetime import datetime
# STORAGE & CATEGORIES
# ============================================================
CATEGORIES = [
    "Food",
    "Transport",
    "Education",
    "Shopping",
    "Entertainment",
    "Hostel",
    "Other"
]
expenses = []
monthly_budget = 0.0
# ============================================================
# MAIN MENU & PROGRAM CONTROL
# ============================================================
def display_menu():
    """
    Display the main menu.
    """
    print("\n")
    print("=" * 50)
    print("       < SMARTSPEND EXPENSE TRACKER >")
    print("=" * 50)
    print("1 -> Add Expense")
    print("2 -> View All Expenses")
    print("3 -> Search Expense")
    print("4 -> Edit Expense")
    print("5 -> Delete Expense")
    print("6 -> Set Monthly Budget")
    print("7 -> View Budget Status")
    print("8 -> Expense Analysis")
    print("9 -> Generate Report")
    print("10 -> Exit")
    print("=" * 50)
def main():
    """
    Main function that controls the entire program.
    """
    print("\n")
    print("=" * 50)
    print("        WELCOME TO SMARTSPEND")
    print("   Student Expense & Budget Tracker")
    print("=" * 50)
    while True:
        display_menu()
        choice = input("Enter your choice: ").strip()
        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            search_expense()
        elif choice == "4":
            edit_expense()
        elif choice == "5":
            delete_expense()
        elif choice == "6":
            set_budget()
        elif choice == "7":
            show_budget_status()
        elif choice == "8":
            display_analysis()
        elif choice == "9":
            generate_report()
        elif choice == "10":
            print("\nThank you for using SmartSpend!")
            print("Program closed successfully.")
            break
        else:
            print("\nInvalid choice.")
            print("Please select a number from 1 to 10.")
# ============================================================
# VALIDATION FUNCTIONS
# ============================================================
def get_positive_amount():
    """
    Get a valid positive amount from the user.
    """
    while True:
        try:
            amount = float(input("Enter amount: ₹"))
            if amount <= 0:
                print("Amount must be greater than zero.")
            else:
                return amount
        except ValueError:
            print("Invalid input. Please enter a valid number.")
def get_date():
    """
    Get a valid date from the user.
    """
    while True:
        date = input("Enter date (DD-MM-YYYY): ").strip()
        try:
            datetime.strptime(date, "%d-%m-%Y")
            return date
        except ValueError:
            print("Invalid date.")
            print("Please use DD-MM-YYYY format.")
def get_non_empty_input(message):
    """
    Get non-empty input from the user.
    """
    while True:
        value = input(message).strip()
        if value:
            return value
        print("Input cannot be empty.")
def get_integer(message):
    """
    Get a valid integer from the user.
    """
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Please enter a valid number.")
# =====================================================
# CATEGORY FUNCTIONS
# =====================================================
def display_categories():
    """
    Display available expense categories.
    """
    print("\n========== EXPENSE CATEGORIES ==========")
    for index, category in enumerate(CATEGORIES, start=1):
        print(f"{index}. {category}")
def select_category():
    """
    Allow the user to select a category.
    """
    display_categories()
    while True:
        try:
            choice = int(input("Select category: "))
            if 1 <= choice <= len(CATEGORIES):
                return CATEGORIES[choice - 1]
            print("Please select a valid category.")
        except ValueError:
            print("Please enter a valid number.")
# ===============================================
# EXPENSE FUNCTIONS
# ===============================================
def generate_expense_id():
    """
    Generate a unique ID for a new expense.
    """
    if not expenses:
        return 1
    return max(expense["id"] for expense in expenses) + 1
def add_expense():
    """
    Add a new expense.
    """
    print("\n=================================")
    print("             ADD EXPENSE")
    print("=================================")
    date = get_date()
    category = select_category()
    description = get_non_empty_input("Enter description: ")
    amount = get_positive_amount()
    expense = {
        "id": generate_expense_id(),
        "date": date,
        "category": category,
        "description": description,
        "amount": amount
    }
    expenses.append(expense)
    print("\nExpense added successfully!")
def display_expense(expense):
    """
    Display a single expense.
    """
    print(
        f"ID: {expense['id']} | "
        f"Date: {expense['date']} | "
        f"Category: {expense['category']} | "
        f"Description: {expense['description']} | "
        f"Amount: ₹{expense['amount']:.2f}"
    )
def view_expenses():
    """
    Display all expenses.
    """
    print("\n=================================")
    print("             ALL EXPENSES")
    print("=================================")
    if not expenses:
        print("No expenses found.")
        return
    for expense in expenses:
        display_expense(expense)
def search_expense():
    """
    Search expenses using category or description.
    """
    print("\n=================================")
    print("            SEARCH EXPENSE")
    print("=================================")

    if not expenses:
        print("No expenses available.")
        return
    keyword = input("Enter category or description to search: ").lower()
    found = False
    for expense in expenses:
        if (
            keyword in expense["category"].lower()
            or keyword in expense["description"].lower()
        ):
            display_expense(expense)
            found = True
    if not found:
        print("No matching expense found.")
def edit_expense():
    """
    Edit an existing expense.
    """
    print("\n=================================")
    print("             EDIT EXPENSE")
    print("=================================")
    if not expenses:
        print("No expenses available.")
        return
    view_expenses()
    expense_id = get_integer("\nEnter expense ID to edit: ")
    for expense in expenses:
        if expense["id"] == expense_id:
            print("\nEnter new details:")
            expense["date"] = get_date()
            expense["category"] = select_category()
            expense["description"] = get_non_empty_input("Enter new description: ")
            expense["amount"] = get_positive_amount()
            print("\nExpense updated successfully!")
            return
    print("Expense ID not found.")
def delete_expense():
    """
    Delete an existing expense.
    """
    print("\n=================================")
    print("            DELETE EXPENSE")
    print("=================================")
    if not expenses:
        print("No expenses available.")
        return
    view_expenses()
    expense_id = get_integer("\nEnter expense ID to delete: ")
    for expense in expenses:
        if expense["id"] == expense_id:
            expenses.remove(expense)
            print("\nExpense deleted successfully!")
            return
    print("Expense ID not found.")
# ============================================================
# BUDGET FUNCTIONS
# ============================================================
def set_budget():
    """
    Set or update the monthly budget.
    """
    global monthly_budget
    print("\n=================================")
    print("          SET MONTHLY BUDGET")
    print("=================================")
    monthly_budget = get_positive_amount()
    print(f"\nMonthly budget set to ₹{monthly_budget:.2f}")
def calculate_total_expense():
    """
    Calculate total amount spent.
    """
    total = 0
    for expense in expenses:
        total += expense["amount"]
    return total
def show_budget_status():
    """
    Display budget and spending status.
    """
    print("\n=================================")
    print("             BUDGET STATUS")
    print("=================================")
    if monthly_budget == 0:
        print("Monthly budget has not been set.")
        return
    total_expense = calculate_total_expense()
    remaining = monthly_budget - total_expense
    print(f"Monthly Budget : ₹{monthly_budget:.2f}")
    print(f"Total Spent    : ₹{total_expense:.2f}")

    if remaining >= 0:
        print(f"Remaining: ₹{remaining:.2f}")
        print("Status: Within budget")
    else:
        print(f"Budget Exceeded: ₹{abs(remaining):.2f}")
        print("Status: Budget exceeded")
# ============================================================
# ANALYSIS FUNCTIONS
# ============================================================
def get_average_expense():
    """
    Calculate average expense.
    """
    if not expenses:
        return 0
    return calculate_total_expense() / len(expenses)
def get_highest_expense():
    """
    Find the highest expense.
    """
    if not expenses:
        return None
    return max(expenses, key=lambda expense: expense["amount"])
def get_lowest_expense():
    """
    Find the lowest expense.
    """
    if not expenses:
        return None
    return min(expenses, key=lambda expense: expense["amount"])
def category_analysis():
    """
    Calculate category-wise spending.
    """
    category_totals = {}
    for expense in expenses:
        category = expense["category"]
        amount = expense["amount"]
        if category not in category_totals:
            category_totals[category] = 0
        category_totals[category] += amount
    return category_totals
def display_analysis():
    """
    Display complete expense analysis.
    """
    print("\n=================================")
    print("EXPENSE ANALYSIS")
    print("=================================")
    if not expenses:
        print("No expenses available for analysis.")
        return
    total = calculate_total_expense()
    average = get_average_expense()
    highest = get_highest_expense()
    lowest = get_lowest_expense()
    print(f"Total Expenses: ₹{total:.2f}")
    print(f"Average Expense : ₹{average:.2f}")
    print("\nHighest Expense")
    print(f"{highest['description']} - ₹{highest['amount']:.2f}")
    print("\nLowest Expense")
    print(f"{lowest['description']} - ₹{lowest['amount']:.2f}")
    print("\nCategory-wise Spending")
    print("-" * 40)
    category_totals = category_analysis()
    for category, amount in category_totals.items():
        print(f"{category:<20} ₹{amount:.2f}")
# ============================================================
# REPORT FUNCTION
# ============================================================
def generate_report():
    """
    Generate a complete expense report.
    """
    print("\n=================================")
    print("       SMARTSPEND MONTHLY REPORT")
    print("=================================")
    if not expenses:
        print("No expenses available.")
        return
    total = calculate_total_expense()
    average = get_average_expense()
    remaining = monthly_budget - total
    print(f"\nMonthly Budget: ₹{monthly_budget:.2f}")
    print(f"Total Spent: ₹{total:.2f}")
    print(f"Average Expense: ₹{average:.2f}")
    if monthly_budget == 0:
        print("Budget Status: Budget not set")
    elif remaining >= 0:
        print(f"Remaining: ₹{remaining:.2f}")
        print("Status: Within budget")
    else:
        print(f"Exceeded By: ₹{abs(remaining):.2f}")
        print("Status: Budget exceeded")
    print("\nCategory-wise Expenses")
    print("-" * 40)
    category_totals = category_analysis()
    for category, amount in category_totals.items():
        print(f"{category:<20} ₹{amount:.2f}")
    print(f"\nNumber of Transactions: {len(expenses)}")
    print("\n=================================")
    print("            END OF REPORT")
    print("=================================")
# ============================================================
# PROGRAM START
# ============================================================
if __name__ == "__main__":
    main()
```
